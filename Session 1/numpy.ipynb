## More NumPy Basics


```python
# basic indexing and slicing
matrix = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
print(matrix[0])
print(matrix[1, 2])
print(matrix[:, 1])
print(matrix[0:2, 1:3])
```

```python
# array math and broadcasting
prices = np.array([100, 200, 300])
tax = prices * 0.05
total = prices + tax
print(prices)
print(tax)
print(total)
```

```python
# reshape a 1D array into 2D
numbers = np.arange(1, 13)
reshaped = numbers.reshape(3, 4)
print(numbers)
print(reshaped)
```

```python
# transpose and flatten
grid = np.array([[1, 2, 3], [4, 5, 6]])
print(grid.T)
print(grid.flatten())
```

```python
# random arrays
np.random.seed(7)
rand_array = np.random.randint(1, 100, size=(3, 3))
print(rand_array)
print(rand_array.min())
print(rand_array.max())
```

```python
scores = np.array([45, 72, 88, 59, 91])
passed = scores[scores >= 60]
print(scores)
print(passed)
```

```python
left = np.array([1, 2, 3])
right = np.array([4, 5, 6])
print(np.concatenate([left, right]))
print(np.vstack([left, right]))
print(np.hstack([left, right]))
```

```python
practice = np.arange(1, 6)
squared = practice ** 2
print(practice)
print(squared)
print(practice.mean())
```

```python
# array statistics
data = np.array([12, 15, 18, 21, 24])
print(data.sum())
print(data.mean())
print(data.min())
print(data.max())
print(data.std())
```

```python
# copy versus view
ori = np.array([1, 2, 3])
view = ori
copy = ori.copy()
view[0] = 99
copy[1] = 77
print(ori)
print(view)
print(copy)
```
